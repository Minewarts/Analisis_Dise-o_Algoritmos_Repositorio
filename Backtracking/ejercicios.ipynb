 ### NOTAS INTRODUCCION AL BACKTRACKING
 Paso por referencia - Estudiar | Bracktrack deshace los pasos que uno hizo antes de saltar | Poda, podar ramas que se saben que no van a servir | Estudiar debuging

```python
from typing import List 
s1= 'adsafd'
s2= 'fas'

def formar_s2 (string1:str, string2:str, current:str = "", index:int = 0)->bool:
    if  current == string2:
        return True
    if len(current) >= len(string2):
        return False
    
    for char in string1: 
        if char is not string2[index]:
            continue
        current+= char 
        if formar_s2(string1, string2, current, index+1):
            return True
        current = current[:-1]

    return False

print(formar_s2(s1,s2))
```

Todas las permutaciones posibles

```python
from typing import List
lista_enteros = [1,2,3,4,5,6,7,8,9]
tamaño_permutacion = 3
def Permutaciones(array:List[int],N:int,permutaciones:List[List[int]] = [], current:List[int] = [])->List[List[int]]:
    if len(current) == N:
        
        permutaciones.append(current.copy())
        return current
    
    for i in range(len(array)):
        if i != 0:
            resto = array[:i] + array[i+1:]
        else:
            resto = array[1:]
        if array[i] in current:
            continue
        current += [array[i]]
        Permutaciones(resto,N,permutaciones,current)
        current.pop()

    return permutaciones
    
        

print (Permutaciones(lista_enteros, tamaño_permutacion))

        

```
