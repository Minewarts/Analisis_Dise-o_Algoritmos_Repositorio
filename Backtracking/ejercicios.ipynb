 Paso por referencia - Estudiar | Bracktrack deshace los pasos que uno hizo antes de saltar | Poda, podar ramas que se saben que no van a servir | Estudiar debuging

```python
from typing import List 
s1= 'adsafd'
s2= 'fas'

def formar_s2 (string1:str, string2:str, current:str = "", index:int = 0)->bool:
    if  current == string2:
        return True
    if len(current) >= len(string2):
        return False
    
    for char in string1: 
        if char is not string2[index]:
            continue
        current+= char 
        if formar_s2(string1, string2, current, index+1):
            return True
        current = current[:-1]

    return False

print(formar_s2(s1,s2))
```

Todas las permutaciones posibles

```python
from typing import List
lista_enteros = [1,2,3,4,5,6,7,8,9]
tamaño_permutacion = 3
def Permutaciones(array:List[int],N:int,permutaciones:List[List[int]] = [], current:List[int] = [])->List[List[int]]:
    if len(current) == N:
        
        permutaciones.append(current.copy())
        return current
    
    for i in range(len(array)):
        if i != 0:
            resto = array[:i] + array[i+1:]
        else:
            resto = array[1:]
        if array[i] in current:
            continue
        current += [array[i]]
        Permutaciones(resto,N,permutaciones,current)
        current.pop()

    return permutaciones
    
        

print (Permutaciones(lista_enteros, tamaño_permutacion))

        

```

generador de sub conjuntos

```python
from typing import List
nums = [1,2,3]

def generador_subconjuntos (array:List[int], current:List[int] = [], subarray:List[List[int]] = [[]])->List[List[int]]:
    
    if len(current) > len(nums):
        return

    if sorted(current) not in subarray:
        subarray.append(current.copy())
    
    for i in range(len(array)): 
        if array[i] in current:
            continue

        current.append(array[i])
        generador_subconjuntos(array,current,subarray)
        current.pop()

    return subarray

print (generador_subconjuntos(nums))
```

remplazar un numero por un elemento menor o igual a k para sumar k 

```python
from typing import List, Tuple

L=[1,2,10,4,1]
k=10
def sumatoria(lista:List[int],target: int, index:int = 0)->Tuple[int, int]:
    if index == len(L)-1:
        return ()
        
        
    for j in range(1,target+1):
        
        aux_lista = lista.copy()
        aux_lista[index] = j
        if sum(aux_lista)> target:
            break
        if sum(aux_lista) == k :
            return lista[index],j

    respuesta = sumatoria(lista, target, index+1)
    if  respuesta != ():
        return respuesta
        
    return ()

print(sumatoria(L, k))
```

se tiene un conjunto de casas, no se puede robar en casas contiguas, dar la mayor cantidad de cantidad robable

```python
from typing import List
casas= [50, 20, 30, 90, 10]

def robar_max_cantidad(h : List[int], current_sum : int = 0, index = 0 )->int:
    if index >= len(h):
        return current_sum

    return max(robar_max_cantidad(h,current_sum + h[index], index+2), robar_max_cantidad(h,current_sum , index+1))

print(robar_max_cantidad(casas))
    
```

Hallar todos los caminos de una matriz de c1 que lleven a c2 y que cuya suma no exceda k

```python
from typing import List
matriz = [
          [2,0,0,2],
          [1,2,0,0],
          [1,0,5,0],
          [5,4,5,2]
          ]
c1 = [1,0]
c2 = [2,3]
k = 7

def Caminos(matriz : List[List[int]], c1 : List[int], c2 : List[int], k : int, i : int = None , j : int = None, sum_act:int = 0,
            camino : List[List[int]] = None, caminos : List [List[List[int]]] = None) -> List [List[List[int]]]:

    if camino is None : camino = []
    if caminos is None: camino = []
    if i is None and j is None:
        i=c1[0]
        j=c1[1]

    sum_act = sum_act + matriz[i][j] 

    if sum_act > k:
        return caminos

    if [i,j] in camino:
        return caminos
    
    if [i,i] == c2:
        camino.append([i,j])
        caminos.append(camino.copy())
        camino.pop()
        return caminos

    
    if i < len(matriz)-1: 
        camino.append([i,j])
        Caminos(matriz,c1,c2,k,i+1,j,sum_act, camino, caminos)
        camino.pop()
    if i > 0: 
        camino.append([i,j])
        Caminos(matriz,c1,c2,k,i-1,j,sum_act, camino, caminos)
        camino.pop()
    if j < len(matriz[0])-1: 
        camino.append([i,j])
        Caminos(matriz,c1,c2,k,i,j+1,sum_act, camino, caminos)
        camino.pop()
    if j > 0: 
        camino.append([i,j])
        Caminos(matriz,c1,c2,k,i,j-1,sum_act, camino, caminos)
        camino.pop()

    return caminos
    

print (Caminos(matriz, c1, c2, k))

```
