 ### NOTAS INTRODUCCION AL BACKTRACKING
 Paso por referencia - Estudiar | Bracktrack deshace los pasos que uno hizo antes de saltar | Poda, podar ramas que se saben que no van a servir | Estudiar debuging

```python
from typing import List 
s1= 'adsafd'
s2= 'fas'

def formar_s2 (string1:str, string2:str, current:str = "", index:int = 0)->bool:
    if  current == string2:
        return True

    
    if len(current) >= len(string2):
        
        return False
    
    for char in string1: 
        if char is not string2[index]:
            continue
        current+= char 
        if formar_s2(string1, string2, current, index+1):
            return True
        current = current[:-1]

    return False

print(formar_s2(s1,s2))
```
