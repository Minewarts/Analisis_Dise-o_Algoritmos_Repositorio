coin_change cantidad minima de monedas para llegar al target?
analizar el mejor escenario posible

```python
#sort(coins) -> O(nlogn)
#max(coins) -> O(n)
from typing import List,Tuple

coins = [1,4,2]
target = 89


def cambio(coins:List[int],target:int, cantidad:int=0, suma:int=0 )-> Tuple[int]:
    coins = sorted(coins, reverse=True)
    resto = target
    for i in range(len(coins)):

        cantidad+=resto//coins[i]
        suma+= coins[i]*(resto//coins[i])
        resto = target-suma

        if resto ==  0:
            break

    return cantidad
print(cambio(coins, target))
```

dada una coleccion de enteros positivos determinar el tamaño maximo del subarreglo cuya sumatoria sea menor o igual a k <= k

```python
lista= [10,1,2,10,1,1,1,1,2,0,4,2,4,2]
k = 5
from typing import List
def calcular_max(lista:List[int],target:int):
    der = 0
    izq = 0
    suma = 0
    maximo = float('-inf')
    flag = True
    while der < len(lista):
        if flag:
            suma += lista[der]
        
        if der != izq and suma > target:
            suma -= lista[izq]
            izq += 1
            flag = False

        if suma <= target and ((der-izq)+1) > maximo :
            maximo = ((der-izq)+1)

        if suma <= target:
            der += 1
            flag = True

        if izq == der and suma > target:
            suma -= lista[izq]
            izq += 1
            der += 1 
            flag = True
            

    return maximo

print (calcular_max(lista,k))
```

Tienes un conjunto de actividades, cada una con una hora de inicio y una hora de finalización. Debes encontrar el número máximo de actividades que una sola persona puede realizar sin que se solapen

```python
inicio = [1, 3, 0, 5, 8, 5]
fin = [2, 4, 6, 7, 9, 9]

start_1= [1, 3, 0, 5, 8, 5] 
finish_1 = [2, 4, 6, 7, 9, 9]


start_2 = [10, 12, 20]
finish_2 = [20, 25, 30]


def choising_activities(inicio:List[int], fin:List[int])->List[int]:
    schedule = []
    for i in range(len(inicio)):
        schedule.append((inicio[i],fin[i]))

    tuplas = sorted(schedule,key = lambda x : x[1])
    print(tuplas)

    contador = 0
    prev=()
    
    for i in range(len(tuplas)):
        if tuplas[i][0] > 24 or tuplas[i][1] >24:
            continue 
        if prev != ():
            if tuplas[i][0]<=prev[1] and tuplas[i][0] >= prev[0]:
                continue
            #if tuplas[i][1]==prev[1] and tuplas[i][1] >= prev[0]:
                #continue
            if tuplas[i][1]>=prev[1] and tuplas[i][0] <= prev[0]:
                continue

        prev = tuplas[i]
        contador += 1
    return contador

    

print(choising_activities(inicio, fin))
print(choising_activities(start_1, finish_1))
print(choising_activities(start_2, finish_2))
```

```python
from typing import List

array = [5,1,4,3]

def unir(array:List[int]):
    cost = 0
    while len(array)>1:
        array = sorted(array)
        num = array.pop(0)
        array [0] = array[0] + num
        cost += array[0]

    return cost

print(unir(array))
```

```python
#heapify(list) ---> convierte uma coleccion(Lista) en heap
#heappush(list) ---> inserta un valor --- logn
#heappop(list, valor) ---> agarra la raiz minimo valor --- logn  ---  hay unas funciones en la 3.14 que nos permiten obtener el maximo valor, con max

import heapq
from typing import List

array = [5, 1, 4, 3]

print(heapq.heapify(array))


def unir_con_heaps(array: List[int]) -> int:
    # 1. Transformamos la lista normal en un Min-Heap
    heapq.heapify(array)
    
    cost = 0
    
    # 2. Mientras haya más de un elemento para sumar
    while len(array) > 1:
        # 3. Sacamos los dos números más pequeños
        primero = heapq.heappop(array)
        segundo = heapq.heappop(array)
        
        # 4. Los sumamos y acumulamos el costo
        suma = primero + segundo
        cost += suma
        
        # 5. Volvemos a meter el resultado de la suma al heap
        heapq.heappush(array, suma)
        
    return cost

print(unir_con_heaps(array)) # Salida: 25
```
