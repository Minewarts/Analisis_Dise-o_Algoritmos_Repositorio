analizar el mejor escenario posible

```python
#sort(coins) -> O(nlogn)
#max(coins) -> O(n)
from typing import List,Tuple

coins = [1,4,2]
target = 89


def cambio(coins:List[int],target:int, cantidad:int=0, suma:int=0 )-> Tuple[int]:
    coins = sorted(coins, reverse=True)
    resto = target
    for i in range(len(coins)):

        cantidad+=resto//coins[i]
        suma+= coins[i]*cantidad
        resto = target-suma

        if resto ==  0:
            break

    return cantidad
print(cambio(coins, target))
```
